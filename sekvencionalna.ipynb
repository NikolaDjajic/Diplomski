```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import gaussian_kde
import numpy as np
from sklearn.utils import resample
```

```python
df = pd.read_csv(r'podaci\clickstream+data+for+online+shopping\e-shop clothing 2008.csv', encoding='cp1252', sep=';')
print(df.columns)
```

```python
# kolone koje želimo u pravilima
cols = [
    'country',
    'page 1 (main category)',
    'page 2 (clothing model)',
    'colour',
    'location',
    'model photography',
    'price 2'
]

df_rules = df[['session ID'] + cols].astype(str)

# dodamo ime kolone ispred vrijednosti
for col in cols:
    df_rules[col] = col + "_" + df_rules[col]

# grupisanje po sesiji
transactions = df_rules.groupby('session ID')[cols].apply(lambda x: sum(x.values.tolist(), []))
```

```python
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)

df_final = pd.DataFrame(te_array, columns=te.columns_)
```

```python
df_final = resample(df_final, n_samples=5, random_state=123)
```

```python
df_final.columns
```

```python
from mlxtend.frequent_patterns import apriori, association_rules

frequent = apriori(df_final, min_support=0.05, use_colnames=True)
print("a")
rules = association_rules(frequent, metric="lift", min_threshold=1.2)

rules = rules.sort_values(by="lift", ascending=False)

print(rules[['antecedents','consequents','support','confidence','lift']])
```
