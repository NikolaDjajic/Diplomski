```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import gaussian_kde
import numpy as np
```

```python
df = pd.read_csv(r'podaci\clickstream+data+for+online+shopping\e-shop clothing 2008.csv', encoding='cp1252', sep=';')
print(df.columns)
```

```python
df = df.sort_values(["session ID", "order"])
```

```python
sequences = df.groupby("session ID")["page 2 (clothing model)"].apply(list)
```

```python
df["event"] = df["page 2 (clothing model)"] + "_" + df["price"].astype(str)

sequences = df.groupby("session ID")["event"].apply(list)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

docs = [" ".join(seq) for seq in sequences]

vectorizer = TfidfVectorizer(ngram_range=(2,2))
X = vectorizer.fit_transform(docs)

print(X.shape)
```
