```python
import pandas as pd
import re

df = pd.read_excel(
    "rezultatiKlasterOSI/rez_PCA.xlsx",
)
df = df['Raspodjela ciljne po klasterima']
```

```python
string_values = df.values.flatten().astype(str)

one_string = ','.join(string_values)

list_of_strings = string_values.tolist()
```

```python
clean=[]
```

```python
for s in list_of_strings:
    numbers = re.findall(r'\d+\.\d+', s)
    clean.append(numbers)
    # print(numbers)
```

```python
clean.pop(5)
```

```python
for c in clean:
    c[1], c[2] = c[2], c[1]
    for i, a in enumerate(c):
        c[i] = round(float(a), 3)
```

```python
rows = []
for numbers in clean:
    rows.append(numbers[:2])  # prva dva broja
    rows.append(numbers[2:])  # druga dva broja
```

```python
df = pd.DataFrame(rows)
df.to_excel("rezultatiKlasterOSI/help/help_PCA.xlsx", index=False, header=False)
```
